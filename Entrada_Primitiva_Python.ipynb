```python
# Entrada de dados primitiva (Teclado) entrada

# É um complemento da saída (Monitor) print
```

```python
nome = input('Digite seu nome: ')
idade = input('Digite seu idade: ')
salario = input('Digite seu salário: ')
print(f'Seu nome é {nome} e sua idade é {idade} e está ganhando {salario}')



```

```python
# Notas 1 e 2 são complementares
# nota 3 única e vai participar da média(n1+n2) + n3 / 2
n1 = int(input('Digite a primeira nota: '))
n2 = int(input('Digite a segunda nota: '))
soma = n1+n2
n3 = int(input('Digite a terceira nota: '))
mf = (soma + n3)  / 2
print(f"A média final é {mf:.2f}")


```

```python
num = int(input('Digite um numero: '))
antecessor =  num - 1
sucessor = num + 1
print(f"O número digitado foi {num}, sendo seu antecessor {antecessor} e seu sucessor {sucessor}")

```
